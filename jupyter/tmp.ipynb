```python
import cv2
import os
from pathlib import Path

def convert_avi_to_mp4_opencv(input_path, output_path=None):
    """
    OpenCV를 사용하여 AVI를 MP4로 변환
    (FFmpeg 없이도 작동)
    """
    input_path = Path(input_path)
    
    if output_path is None:
        output_path = input_path.with_suffix('.mp4')
    else:
        output_path = Path(output_path)
    
    print(f"변환 시작: {input_path} -> {output_path}")
    
    try:
        # 비디오 캡처 객체 생성
        cap = cv2.VideoCapture(str(input_path))
        
        if not cap.isOpened():
            print(f"비디오 파일을 열 수 없습니다: {input_path}")
            return False
        
        # 비디오 속성 가져오기
        fps = int(cap.get(cv2.CAP_PROP_FPS))
        width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
        
        print(f"비디오 정보: {width}x{height}, {fps}fps")
        
        # 비디오 라이터 설정
        fourcc = cv2.VideoWriter_fourcc(*'mp4v')
        out = cv2.VideoWriter(str(output_path), fourcc, fps, (width, height))
        
        frame_count = 0
        while True:
            ret, frame = cap.read()
            if not ret:
                break
            
            out.write(frame)
            frame_count += 1
            
            if frame_count % 100 == 0:
                print(f"프레임 처리 중: {frame_count}")
        
        # 리소스 해제
        cap.release()
        out.release()
        
        print(f"✅ 변환 완료: {output_path}")
        print(f"총 프레임 수: {frame_count}")
        return True
        
    except Exception as e:
        print(f"❌ 변환 실패: {e}")
        return False

# 사용 예시
input_file = "media/videos/standard/2_1_20250128143052_clean.avi"
success = convert_avi_to_mp4_opencv(input_file)
```

```python
import cv2
import os
from pathlib import Path
import subprocess
import sys

def check_ffmpeg():
    """FFmpeg 설치 확인"""
    try:
        result = subprocess.run(['ffmpeg', '-version'], capture_output=True, text=True)
        if result.returncode == 0:
            version_line = result.stdout.split('\n')[0]
            print(f"✅ FFmpeg 설치됨: {version_line}")
            return True
        else:
            print("❌ FFmpeg 설치에 문제가 있습니다.")
            return False
    except FileNotFoundError:
        print("❌ FFmpeg가 설치되지 않았습니다.")
        print("설치 방법:")
        print("1. Conda: conda install -c conda-forge ffmpeg")
        print("2. 직접 다운로드: https://www.gyan.dev/ffmpeg/builds/")
        return False

def check_video_info(video_path):
    """비디오 파일 정보 확인"""
    video_path = Path(video_path)
    
    if not video_path.exists():
        print(f"❌ 파일이 존재하지 않습니다: {video_path}")
        return None
    
    cap = cv2.VideoCapture(str(video_path))
    
    if not cap.isOpened():
        print(f"❌ 파일을 열 수 없습니다: {video_path}")
        return None
    
    # 비디오 정보 추출
    fps = cap.get(cv2.CAP_PROP_FPS)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    duration = frame_count / fps if fps > 0 else 0
    
    # 코덱 정보
    fourcc = int(cap.get(cv2.CAP_PROP_FOURCC))
    codec = "".join([chr((fourcc >> 8 * i) & 0xFF) for i in range(4)])
    
    # 파일 크기
    file_size = video_path.stat().st_size / (1024 * 1024)
    
    print(f"📹 {video_path.name}")
    print(f"   해상도: {width}x{height}")
    print(f"   FPS: {fps:.2f}")
    print(f"   재생시간: {duration:.1f}초")
    print(f"   코덱: {codec}")
    print(f"   크기: {file_size:.1f}MB")
    
    cap.release()
    return True

def convert_to_chrome_mp4(input_path, output_path=None):
    """Chrome 호환 MP4로 변환"""
    input_path = Path(input_path)
    
    if not input_path.exists():
        print(f"❌ 입력 파일 없음: {input_path}")
        return False
    
    # 출력 파일명 생성
    if output_path is None:
        # AVI 파일인 경우 확장자를 MP4로 변경
        if input_path.suffix.lower() == '.avi':
            output_path = input_path.parent / f"{input_path.stem}.mp4"
        else:
            output_path = input_path.parent / f"{input_path.stem}_chrome.mp4"
    else:
        output_path = Path(output_path)
    
    print(f"🔄 변환: {input_path.name} → {output_path.name}")
    
    # FFmpeg 명령어 (Chrome 최적화)
    cmd = [
        'ffmpeg',
        '-i', str(input_path),
        '-c:v', 'libx264',           # H.264 코덱
        '-profile:v', 'baseline',    # 기본 프로파일
        '-level', '3.0',             # 호환 레벨
        '-pix_fmt', 'yuv420p',       # 픽셀 포맷
        '-c:a', 'aac',               # AAC 오디오
        '-b:a', '128k',              # 오디오 비트레이트
        '-movflags', '+faststart',   # 웹 최적화
        '-preset', 'fast',           # 빠른 인코딩
        '-crf', '23',                # 품질
        '-y',                        # 덮어쓰기
        str(output_path)
    ]
    
    try:
        result = subprocess.run(cmd, capture_output=True, text=True)
        
        if result.returncode == 0:
            if output_path.exists():
                original_size = input_path.stat().st_size / (1024 * 1024)
                new_size = output_path.stat().st_size / (1024 * 1024)
                print(f"✅ 변환 완료!")
                print(f"   원본: {original_size:.1f}MB")
                print(f"   변환: {new_size:.1f}MB")
                return True
            else:
                print("❌ 변환된 파일이 생성되지 않았습니다.")
                return False
        else:
            print(f"❌ 변환 실패: {result.stderr}")
            return False
            
    except Exception as e:
        print(f"❌ 오류 발생: {e}")
        return False

def convert_project_videos():
    """프로젝트의 모든 비디오 파일 변환 (AVI + MP4)"""
    print("🎬 프로젝트 비디오 변환 시작")
    print("=" * 60)
    
    # FFmpeg 확인
    if not check_ffmpeg():
        return
    
    # 변환할 디렉토리
    video_dirs = [
        "../media/videos/standard",
        "../media/videos/user"
    ]
    
    total_converted = 0
    
    for video_dir in video_dirs:
        video_path = Path(video_dir)
        
        if not video_path.exists():
            print(f"⚠️ 디렉토리 없음: {video_dir}")
            continue
        
        print(f"\n📁 {video_dir} 처리 중...")
        
        # AVI와 MP4 파일 모두 찾기
        video_files = list(video_path.glob("*.avi")) + list(video_path.glob("*.mp4"))
        
        if not video_files:
            print("   변환할 파일이 없습니다.")
            continue
        
        # AVI 파일 우선 처리
        avi_files = [f for f in video_files if f.suffix.lower() == '.avi']
        mp4_files = [f for f in video_files if f.suffix.lower() == '.mp4']
        
        print(f"   발견된 파일: AVI {len(avi_files)}개, MP4 {len(mp4_files)}개")
        
        # AVI 파일들을 MP4로 변환
        for avi_file in avi_files:
            print(f"\n📹 AVI → MP4 변환:")
            if check_video_info(avi_file):
                print(f"\n🔄 변환 시작:")
                if convert_to_chrome_mp4(avi_file):
                    total_converted += 1
                print("-" * 40)
        
        # MP4 파일들을 Chrome 호환으로 변환
        for mp4_file in mp4_files:
            print(f"\n📹 MP4 → Chrome 호환 MP4 변환:")
            if check_video_info(mp4_file):
                print(f"\n🔄 변환 시작:")
                if convert_to_chrome_mp4(mp4_file):
                    total_converted += 1
                print("-" * 40)
    
    print(f"\n🎉 변환 완료! 총 {total_converted}개 파일 변환됨")

def update_database_paths():
    """데이터베이스 경로 업데이트 (AVI → MP4)"""
    print("\n🗄️ 데이터베이스 경로 업데이트 안내")
    print("=" * 50)
    print("변환 완료 후 다음 SQL을 실행하세요:")
    print()
    print("UPDATE videos")
    print("SET video_path = REPLACE(video_path, '.avi', '.mp4')")
    print("WHERE video_path LIKE '%.avi';")
    print()
    print("변경 결과 확인:")
    print("SELECT * FROM videos;")

def test_converted_videos():
    """변환된 파일들 테스트"""
    print("\n🧪 변환된 파일 테스트")
    print("=" * 40)
    
    video_dirs = ["media/videos/standard", "media/videos/user"]
    
    for video_dir in video_dirs:
        video_path = Path(video_dir)
        
        if video_path.exists():
            # 변환된 MP4 파일들 찾기
            mp4_files = list(video_path.glob("*.mp4"))
            
            if mp4_files:
                print(f"\n📁 {video_dir}:")
                for mp4_file in mp4_files:
                    print(f"✅ {mp4_file.name}")
            else:
                print(f"\n📁 {video_dir}: 변환된 파일 없음")

# 메인 실행
if __name__ == "__main__":
    print("🚀 비디오 변환 도구 (AVI + MP4 지원)")
    print("=" * 60)
    
    # 1. 프로젝트 비디오 변환
    convert_project_videos()
    
    # 2. 데이터베이스 업데이트 안내
    update_database_paths()
    
    # 3. 변환 결과 테스트
    test_converted_videos()
    
    print("\n✨ 완료! 이제 웹에서 비디오를 테스트해보세요.")
```
