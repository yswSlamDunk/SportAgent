### 장기 메모리 구조
1. Preferences(선호 답변 스타일)
    * reply_length(답변 길이): short / medium / long / summary
    * tone(톤): friendly / professional / blunt / strict
    * format_preference(선호 형식): bullet / steps / table / tags
    * language_style(말투): 존댓말 / 반말 / 자연스러운
    * explanation_depth(기술 설명 난이도): beginner / intermediate / expert
2. User Info(유저 정보)
    1) 기본 정보
        * age (나이)
        * gender (성별)
        * activity_type(운동 수준): 일반인 / 동호인 / 선수
        * sport_experience(운동 경력) 
    2) 신체 정보
        * 부상정보
            * injury_history(부상이력)
            * movement_limitations(움직임 제한)
        * 신체적 특성
            * hight (키)
            * weight (몸무게)
            * other_characteristics (기타 신체적 특징)
3. Fact(학습된 지식)
4. Temp(임시)
5. Trigger(트리거로 인해 저장된 내용)
    * 장기 메모리 갱신 과정에서 구조화 됨
    * 구조화 되지 못한다면 임시에 저장됨
